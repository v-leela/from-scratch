```python
import pandas as pd
import numpy as np
```

## Load the data

```python
insurance = pd.read_csv(os.path.join(project_dir, "data/insurance.csv"))
```

```python
insurance = pd.get_dummies(insurance, columns=["region"], dtype=int)
insurance["sex"] = insurance["sex"].map({"male": 1, "female": 0})
insurance["smoker"] = insurance["smoker"].map({"yes": 1, "no": 0})
```

```python
X = insurance.drop(columns="charges")
y = insurance["charges"]
```

```python
n = len(X)
```

```python
indices = np.arange(n)
np.random.shuffle(indices)
```

```python
s1 = int(0.75 * n)
```

```python
X_train = X.iloc[indices[:s1],]
y_train = y.iloc[indices[:s1],]
X_test = X.iloc[indices[s1:],]
y_test = y.iloc[indices[s1:],]
```

## Linear Regression

```python
lr.r_score(X_test, y_test)
```

```python
lr.rmse(X_test, y_test)
```
