## Import the Classes

```python
from models.decisiontreeclassifier import DecisionTreeClassifier
from models.randomforestclassifier import RandomForestClassifier
from models.xgboostclassifier import XGBoostClassifier
from models.logisticregression import LogisticRegression
```

```python
import pandas as pd
import numpy as np
```

## Load the data

```python
heart = pd.read_csv(os.path.join(project_dir, "data/heart.csv"))
```

```python
X = heart.drop(columns="target")
y = heart["target"]
```

```python
n = len(X)
```

```python
indices = np.arange(n)
np.random.shuffle(indices)
```

```python
s1 = int(0.75 * n)
```

```python
X_train = X.iloc[indices[:s1],]
y_train = y.iloc[indices[:s1],]
X_test = X.iloc[indices[s1:],]
y_test = y.iloc[indices[s1:],]
```

## Logistic Regression

```python
lg = LogisticRegression()
```

```python
lg.fit(X_train, y_train)
```

```python
lg.accuracy(X_test, y_test)
```

## Decision Tree

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(X_train, y_train)
```

```python
dtc.accuracy_test(X_test, y_test)
```

## Random Forest

```python
rfc = RandomForestClassifier()
```

```python
rfc.fit(X_train, y_train)
```

```python
rfc.accuaracy_test(X_test, y_test)
```

## XGBoost

```python
xgbc = XGBoostClassifier()
```

xgb takes train and val at the same time for fit

```python
s1 = int(0.7 * n)
s2 = int(0.85 * n)
```

```python
X_train = X.iloc[indices[:s1],]
y_train = y.iloc[indices[:s1],]
X_val = X.iloc[indices[s1:s2],]
y_val = y.iloc[indices[s1:s2],]
X_test = X.iloc[indices[s2:],]
y_test = y.iloc[indices[s2:],]
```

```python
xgbc.fit(X_train, y_train, X_val, y_val)
```

```python
xgbc.accuracy(X_test, y_test)
```
